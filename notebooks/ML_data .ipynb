```python
import pandas as pd
```

```python
path = r"C:\Users\SVS\Desktop\ML Project\data\processed"

apy = pd.read_csv(path + r"\APY_Telangana.csv")
rainfall = pd.read_csv(path + r"\Rainfall_Telangana.csv")
irrigation = pd.read_csv(path + r"\Crop_Irrigation_Telangana.csv")
irrigation_sources = pd.read_csv(path + r"\Irrigation_Sources_Telangana.csv")
```

```python
model_districts = [
    "Adilabad",
    "Karimnagar",
    "Khammam",
    "Mahabubnagar",
    "Medak",
    "Nalgonda",
    "Nizamabad",
    "Ranga Reddy",
    "Warangal"
]
```

```python
model_years = [
    "2014-2015",
    "2015-2016",
    "2016-2017",
    "2017-2018",
    "2018-2019",
    "2019-2020",
    "2020-2021",
    "2021-2022",
    "2022-2023"
]
```

```python
## area production yield
apy_ml = apy[
    (apy["old_district_name"].isin(model_districts)) &
    (apy["year"].isin(model_years))
].copy()
```

```python
print(apy_ml.shape)
```

```python
print(apy_ml["old_district_name"].nunique())
print(sorted(apy_ml["old_district_name"].unique()))
```

```python
## crop irrigation
irrigation_ml = irrigation[
    (irrigation["old_district_name"].isin(model_districts)) &
    (irrigation["year"].isin(model_years))
].copy()
```

```python
print(irrigation_ml.shape)
```

```python
## irrigation sources 
irrigation_sources_ml = irrigation_sources[
    (irrigation_sources["old_district_name"].isin(model_districts)) &
    (irrigation_sources["year"].isin(model_years))
].copy()
```

```python
print(irrigation_sources_ml.shape)
```

```python
print("APY:", apy_ml.shape)
print("Irrigation:", irrigation_ml.shape)
print("Irrigation Sources:", irrigation_sources_ml.shape)
```

```python
rainfall["date"] = pd.to_datetime(
    rainfall["date"],
    errors="coerce"
)
```

```python
print(rainfall["date"].dtype)
```

```python
rainfall["date"] = pd.to_datetime(
    rainfall["date"],
    errors="coerce"
)

rainfall["agri_year"] = rainfall["date"].dt.year.astype(str)

rainfall.loc[
    rainfall["date"].dt.month >= 6,
    "agri_year"
] = (
    rainfall.loc[
        rainfall["date"].dt.month >= 6,
        "date"
    ].dt.year.astype(str)
    + "-"
    + (
        rainfall.loc[
            rainfall["date"].dt.month >= 6,
            "date"
        ].dt.year + 1
    ).astype(str)
)

rainfall.loc[
    rainfall["date"].dt.month < 6,
    "agri_year"
] = (
    (
        rainfall.loc[
            rainfall["date"].dt.month < 6,
            "date"
        ].dt.year - 1
    ).astype(str)
    + "-"
    + rainfall.loc[
        rainfall["date"].dt.month < 6,
        "date"
    ].dt.year.astype(str)
)
```

```python
rainfall_ml = rainfall[
    (rainfall["district_name"].isin(model_districts)) &
    (rainfall["agri_year"].isin(model_years))
].copy()
```
