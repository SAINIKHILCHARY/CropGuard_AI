```python
import pandas as pd
```

```python
path = r"C:\Users\SVS\Desktop\ML Project\data\raw"

apy = pd.read_csv(path + r"\crop-wise-area-production-yield.csv")
rainfall = pd.read_csv(path + r"\daily-rainfall-data-district-level.csv")
irrigation = pd.read_csv(path + r"\crop-wise-irrigated-area.csv")
irrigation_sources = pd.read_csv(path + r"\sources-of-irrigation.csv")
soil = pd.read_csv(path + r"\soil-nutrient-analysis.csv")
```

```python
apy_tg = apy[ apy["state_name"].str.strip().str.lower() == "telangana"].copy()

rainfall_tg = rainfall[ rainfall["state_name"].str.strip().str.lower() == "telangana"].copy()

irrigation_tg = irrigation[ irrigation["state_name"].str.strip().str.lower() == "telangana"].copy()

irrigation_sources_tg = irrigation_sources[  irrigation_sources["state_name"].str.strip().str.lower() == "telangana"].copy()

soil_tg = soil[soil["state_name"].str.strip().str.lower() == "telangana"].copy()
```

```python
print("APY:", apy_tg.shape)
print("Rainfall:", rainfall_tg.shape)
print("Irrigation:", irrigation_tg.shape)
print("Irrigation Sources:", irrigation_sources_tg.shape)
print("Soil:", soil_tg.shape)
```

```python
all_districts = sorted(apy_tg["district_name"].unique())

unmapped = [
    district for district in all_districts
    if district not in district_mapping
]

print("Total districts:", len(all_districts))
print("Mapped districts:", len(all_districts) - len(unmapped))
print("Unmapped districts:", unmapped)
```

```python
mapping_df = pd.DataFrame(
    list(district_mapping.items()),
    columns=["district_name", "old_district_name"]
)

mapping_df
```

```python
## adding new column to the datasets 

apy_tg["old_district_name"] = apy_tg["district_name"].map(district_mapping)

irrigation_tg["old_district_name"] = irrigation_tg["district_name"].map(district_mapping)

irrigation_sources_tg["old_district_name"] = irrigation_sources_tg["district_name"].map(district_mapping)

soil_tg["old_district_name"] = soil_tg["district_name"].map(district_mapping)
```

```python
## saving the processes datasets to the proceses folder 
processed_path = r"C:\Users\SVS\Desktop\ML Project\data\processed"

apy_tg.to_csv(
    processed_path + r"\APY_Telangana.csv",
    index=False
)

rainfall_tg.to_csv(
    processed_path + r"\Rainfall_Telangana.csv",
    index=False
)

irrigation_tg.to_csv(
    processed_path + r"\Crop_Irrigation_Telangana.csv",
    index=False
)

irrigation_sources_tg.to_csv(
    processed_path + r"\Irrigation_Sources_Telangana.csv",
    index=False
)

soil_tg.to_csv(
    processed_path + r"\Soil_Telangana.csv",
    index=False
)

print("All Telangana datasets saved successfully.")
```
